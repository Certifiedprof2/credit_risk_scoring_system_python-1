## Task 2: Calculate the Debt-to-Income Ratio
### Calculate the percentage of the applicant's monthly income that goes toward loan repayment.

```python
   # Debt-to-Income Ratio = (Monthly Loan Repayment ÷ Monthly Income) × 100

monthly_loan_repayment = 120000
monthly_income = 450000
percentage = 100
debt_to_income_ratio = (monthly_loan_repayment / monthly_income) * percentage
print("Debt to income ratio:", debt_to_income_ratio)
```

## Task 3: Calculate Total Loan Exposure
#### The bank wants to know the customer's total loan exposure if the new loan is approved.

```python
#  Total Loan Exposure = Existing Loan Balance + Requested Loan Amount

existing_loan_balance = 850000
requested_loan_amount = 1500000
total_loan_exposure = existing_loan_balance + requested_loan_amount

print("Total loan exposure:", total_loan_exposure)
```
